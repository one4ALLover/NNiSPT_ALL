```python
# =========================
# INSTALL (НЕ ТРОГАЕМ requests)
# =========================
!pip install -q transformers accelerate
!pip install -q langchain langchain-community langchain-huggingface

# =========================
# MODEL
# =========================
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
from langchain_huggingface import HuggingFacePipeline

model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, device_map="auto")

pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=128,
    do_sample=False
)

llm = HuggingFacePipeline(pipeline=pipe)

# =========================
# TOOL (обязательно docstring)
# =========================
from langchain.tools import tool
from datetime import datetime

@tool
def current_date() -> str:
    """Возвращает текущую дату"""
    return datetime.now().strftime("%Y-%m-%d")

tools = [current_date]

# =========================
# RAG (без RetrievalQA)
# =========================
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import InMemoryVectorStore

with open("article.txt", "w", encoding="utf-8") as f:
    f.write("Yandex is a Russian technology company known for its AI technologies.")

docs = TextLoader("article.txt").load()
chunks = CharacterTextSplitter(chunk_size=300).split_documents(docs)

emb = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
db = InMemoryVectorStore.from_documents(chunks, emb)

def rag(query):
    docs = db.similarity_search(query)
    context = "\n".join([d.page_content for d in docs])
    return llm.invoke(f"Context: {context}\nQuestion: {query}")

# =========================
# AGENT (НОВАЯ СТАБИЛЬНАЯ МОДЕЛЬ)
# =========================
from langchain_core.messages import HumanMessage, SystemMessage

SYSTEM_PROMPT = """
Ты ассистент.
Если вопрос про дату — используй tool current_date.
Если не нужен tool — отвечай сам.
"""

def agent(question: str):
    # простая логика tool-calling вручную (СТАБИЛЬНО)
    if "дата" in question.lower():
        return current_date.invoke({})
    return llm.invoke(question)

# =========================
# TEST 1 (RAG)
# =========================
print("\n===== TEST 1 =====")
print(rag("Какая компания известна своими ИИ технологиями?"))

# =========================
# TEST 2 (AGENT)
# =========================
print("\n===== TEST 2 =====")
print(agent("Какая сегодня дата?"))
```
